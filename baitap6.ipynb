```python
#W6A1
a = input().split()
b = []
for x in a:
    if x not in b:
        b.append(x)
print(b)
```

```python
#W6A2
a = list(map(int, input().split()))
b = []
s = 0
for x in a:
    s += x
    b.append(s)
print(b)
```

```python
#W6A3
a = tuple(input().split())
k = int(input())
k %= len(a)
print(a[k:] + a[:k])
```

```python
#W6A4
a = input().split()
d = {}
for x in a:
    k, v = x.split(':')
    d.setdefault(k, []).append(v)
print(d)
```

```python
#W6A5
a = list(map(int, input().split()))
p = n = z = 0
for x in a:
    if x > 0: p += 1
    elif x < 0: n += 1
    else: z += 1
print({'positives': p, 'negatives': n, 'zeros': z})
```

```python
#W6A7
a = tuple(input().split())
print(a[0], a[-1], a[::-1])
```

```python
#W6A8
a = input().split()
d = {}
for x in a:
    d[x] = d.get(x, 0) + 1
print(d)
```
