```python
#W6A1
a = list(map(int,input().split()))
b = []
for x in a:
    if x not in b:
        b.append(x)
print(b)
```

```python
#W6A2
a = list(map(int, input().split()))
b = []
s = 0
for x in a:
    s += x
    b.append()
print(b)
```

```python
#W6A3
s = input()
tup_list = [int(x) for x in s.split()]
tup = tuple(tup_list)
k = int(input())
n = len(tup)
k = k % n  
rotated_tup = tup[k:] + tup[:k]
print(tup)
print(rotated_tup)
```

```python
#W6A4
a = input().split()
d = {}
for x in a:
    k, v = x.split(':')
    d.setdefault(k, []).append(v)
print(d)
```

```python
#lại
input_str = input()
pairs = input_str.split()
result = {}
for pair in pairs:
    key, value = pair.split(":")
    if key in result:
        result[key].append(value)
    else:
        result[key] = [value]
print(result)
```

```python
#W6A5
a = list(map(int, input().split()))
p = n = z = 0
for x in a:
    if x > 0: p += 1
    elif x < 0: n += 1
    else: z += 1
print({'positives': p, 'negatives': n, 'zeros': z})
```

```python
#W6A9
def merge_and_sort_dicts(dict1, dict2):
    merged_dict = dict1.copy()
    merged_dict.update(dict2) 
    merged_dict.update(dict1) 
    sorted_items = sorted(merged_dict.items())
    sorted_dict = dict(sorted_items)

    print(sorted_dict)
dict_a = {'a': 1, 'b': 2, 'c': 3}
dict_b = {'b': 20, 'd': 4, 'e': 5}

merge_and_sort_dicts(dict_a, dict_b)
```
