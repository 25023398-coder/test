```python
#W5A3
try:
    x = int(input())
    if x > 1:
        for i in range(2,x):
            if x % i == 0:
                print(x,"không là snt")
                break
        else:
            print(x,"là snt")
    else:
        print(x,"không là snt")
except ValueError:
    print("Lỗi")
```

```python
#W5A4
try:
    def perfect(a:int) -> bool:
        if a<=1:
            return False
        total = 0
        for i in range(1,a):
            if a%i == 0:
                total += i
        return total == a
    x = int(input())
    if perfect(x):
        print(x,"là số hoàn hảo")
    else:
        print(x,"không phải số hoàn hảo")
except ValueError:
    print("Lỗi")
```

```python
#W5A4 (cách khác)
try:
    def perfect(a):
        return sum(i for i in range(1, a) if a % i == 0) == a
    x = int(input())
    print("Hoàn hảo" if perfect(x) else "Không hoàn hảo")
except ValueError:
    print("Lỗi")
```

```python
#W5A5
try:
    def find(lst,k):
        for i in range(len(lst)):
            if lst[i]==k:
                return i
        return -1
    ds=[1,2,3,4,5,6,7]
    k = int(input())
    print(find(ds,k))
except ValueError:
    print("Lỗi")
```

```python
#W5A6
def giai_thua(n):
    if n < 0:
        return False
    tich = 1
    for i in range(1, n + 1):
        tich *= i
    return tich
try:
    n = int(input())
    print(f"{n}! =", giai_thua(n))
except ValueError:
    print("Lỗi")
```
