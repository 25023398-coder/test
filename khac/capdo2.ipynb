```python
#Viết một chương trình chuyển đổi nhiệt độ từ độ C sang độ F hoặc ngược lại, dựa vào lựa chọn của người dùng.
def c_to_f(c):
    return c*1.8+32
def f_to_c(f):
    return (f-32)/1.8
print("CHỌN KIỂU CHUYỂN ĐỔI")
print("1.Đổi từ độ C sang độ F")
print("2.Đổi từ độ F sang độ C")
lua_chon = int(input())
if lua_chon == 1:
    do_c = float(input())
    do_f = c_to_f(do_c)
    print(f"{do_c} = {do_f}")
elif lua_chon == 2:
    do_f = float(input())
    do_c = f_to_c(do_f)
    print(f"{do_f} = {do_c}")
else:
    print("Lỗi")
```

```python
#Viết một hàm để xoay một ma trận 90 độ theo chiều kim đồng hồ.
"""
hàng đầu tiên thành cột cuối cùng
hàng giữa thành cột giữa
hàng cuối cùng thành cột đầu tiên
"""
def xoaymatran(matrix):
    hang = len(matrix)  #Đếm số hàng trong ma trận, tức là số danh sách con, ở đây là 3ds
    cot = len(matrix[0])
"""
tại sao lại là matrix[0]?
matrix là dsach 2 chiểu (dsach của dsach con) -> matrix[0] là lấy hàng đầu tiên [1,2,3]
len(matrix(0)) đếm số phần tử trong hàng đầu tiên -> chính là đếm số cột
"""
    ket_qua = [[0] * hang for _ in range(cot)] #tạo ma trận mới để điền dữ liệu sau khi xoay
"""
[0]*hang: tạo ra 1 dsach gồm hang số 0
   ex: hang=3 -> tạo [0,0,0]
for _ in range(cot): lặp lại việc tạo dsach [0,0,0] bằng số cột
-> kết quả là ma trận rỗng toàn số 0
"""
    for i in range(hang):   #duyệt từng hàng
        for j in range(cot):  #duyệt từng cột
            ket_qua[j][hang - 1 - i] = matrix[i][j]
    return ket_qua
ma_trix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
ma_trix_xoay = xoaymatran(ma_trix)
for dong in ma_trix_xoay:
    print(dong)
```

```python
#Viết một chương trình để kiểm tra xem một chuỗi con có tồn tại trong chuỗi khác hay không.
def kiemtrachuoi(a,b):
    if a in b:
        return "chuỗi con nằm trong chuỗi lớn"
    else:
        return "chuỗi con không nằm trong chuỗi lớn"
chuoicon = input()
chuoilon = input()
print(kiemtrachuoi(chuoicon,chuoilon))
```

```python
#Viết một hàm để tìm số lớn thứ hai trong một danh sách số nguyên
def solonthuhai(danhsach):
    if len(danhsach)<2:
        return "lỗi"
    solonnhat = max(danhsach[0],danhsach[1])       #so sánh 2 số đầu tiên của danh sách, đặt làm max,min
    solonthuhai = min(danhsach[0],danhsach[1])
    for so in danhsach[2:]:
        if so > solonnhat:
            solonthuhai = solonnhat
            solonnhat = so
        elif so > solonthuhai and so < solonnhat:
            solonthuhai = so
    return solonthuhai
danh_sach = list(map(int,input().split()))
ket_qua = solonthuhai(danh_sach)
print(ket_qua)
```

```python
#tương tự nhưng tìm số bé thứ hai
def sobethuhai(danhsach):
    if len(danhsach) < 2:
        return "lỗi"
    sobenhat = min(danhsach[0],danhsach[1])
    sobethuhai = max(danhsach[0],danhsach[1])
    for so in danhsach:
        if so < sobenhat:
            sobethuhai = sobenhat
            sobenhat = so
        elif so < sobethuhai and so > sobenhat:
            sobethuhai = so
    return sobethuhai
danh_sach = list(map(int,input().split()))
ket_qua = sobethuhai(danh_sach)
print(ket_qua)
```

```python
#Viết một hàm để tính giai thừa của một số nguyên dương
import math
def tinhgiaithua(n):
    if n % 2 == 0:
        return math.factorial(n)
    else:
        return "lỗi"
a = int(input())
print(tinhgiaithua(a))
```

```python
#vẫn bài trên nhưng không dùng math
def tinhgiaithua(n):
    if n == 0 or n == 1:
        return 1
    elif n > 1:
        return n*tinhgiaithua(n-1)
    else:
        return "lỗi"
a = int(input())
print(tinhgiaithua(a))
```
