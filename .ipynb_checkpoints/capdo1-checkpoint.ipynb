```python
#Viết chương trình kiểm tra xem một số có phải là số nguyên tố hay không
#Số nguyên tố là các số tự nhiên lớn hơn 1, chỉ có đúng hai ước số dương là 1 và chính nó
x = int(input())
is_prime =True
for i in range(2,x):
    if x%i==0:
        is_prime = False
        break
if is_prime:
    print("là số nguyên tố")
else:
    print("không là số nguyên tố")
```

```python
#vẫn là bài trên nhưng làm lại cho nhớ
x = int(input())
is_prime=True
for i in range(2,x):
    if x%i==0:
        is_prime=False
        break
if is_prime:
    print("là số nguyên tố")
else:
    print("không là số nguyên tố")
```

```python
#Viết chương trình in ra tất cả các số nguyên tố nhỏ hơn 100
for i in range(2,100):
    is_prime = True
    for j in range(2,i):
        if i%j==0:
            is_prime=False
            break
    if is_prime:
        print(i)
```

```python
#vẫn là bài trên nhưng dùng cách gọi hàm, dài dòng hơn chút
def giaithua(n):
    if n == 0:
        return 1
    else:
        return n*giaithua(n-1)
x = int(input())
print(giaithua(x))
```

```python
#lại lần nữa vì vẫn hơi lú
def giaithua(n):
    if n == 0:
        return 1
    else:
        return n*giaithua(n-1)
x = int(input())
print(giaithua(x))
```

```python
#So sánh hai số
def sosanh(x,y):
    if x>y:
        return f"{x} lớn hơn {y}"
    elif x<y:
        return f"{x} nhỏ hơn {y}"
    else:
        return f"{x} bằng {y}"
a = int(input())
b = int(input())
print(sosanh(a,b))
```

```python
#ìm số chẵn đầu tiên trong một danh sách
danhsach = [1,2,3,4,5,6,7,8,9]
def timsochan(danhsach):
    for i in danhsach:
        if i%2==0:
            return i
#Đoạn này mình cũng ko rõ lắm tại sao return i lại không phải tất cả các số chẵn mà lại chỉ số đầu tiên

    return "không có số chẵn"
print(timsochan(danhsach))
```

```python
#thử lại bài trên
ds = [1,2,3,4,5,6]
def sochan(ds):
    for i in ds:
        if i%2==0:
            return i
    return "none"
print(sochan(ds))
```

```python
#sửa lại đề bài trên 1 chút
#in tất cả số chẵn
danhsach=[1,2,3,4,5,6]
so_chan=[]
def sochan(danhsach): #này là mới định nghĩa hàm
    for i in danhsach:
        if i%2==0:
            so_chan.append(i)
    return "none"
sochan(danhsach) #này là gọi hàm để truyền giá trị vào trong hàm
print(so_chan)
```

```python
#bài trên nhưng dài hơn
danhsach=[123,456,789]
solonnhat=danhsach[0]
for i in danhsach:
    if i>solonnhat:
        solonnhat = i
print(solonnhat)
```

```python
#In ra các số nguyên tố từ 2 đến 100, làm rồi nhưng thử lại
for i in range(2,100):
    is_prime=True
    for j in range(2,i):
        if i%j==0:
            is_prime=False
            break
    if is_prime:
        print(i)
```
