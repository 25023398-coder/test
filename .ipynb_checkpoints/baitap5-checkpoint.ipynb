```python
#W5A1
try:
    def Max(a:int, b:int) -> int:
        if a>b:
            return(a)
        else:
            return(b)
    x = int(input())
    y = int(input())
    print(Max(x, y))
except ValueError:
    print("Lỗi")
```

```python
#W5A2
try:
    def swap(a:int, b:int) -> [int,int]:
        return b,a
    x = int(input())
    y = int(input())
    print(swap(x,y))
except ValueError:
    print("Lỗi")
```

```python
#W5A3
try:
    x = int(input())
    if x > 1:
        for i in range(2,x):
            if x % i == 0:
                print(x,"không là snt")
                break
        else:
            print(x,"là snt")
    else:
        print(x,"không là snt")
except ValueError:
    print("Lỗi")
```

```python
#W5A4
try:
    def perfect(a:int) -> bool:
        if a<=1:
            return False
        total = 0
        for i in range(1,a):
            if a%i == 0:
                total += i
        return total == a
    x = int(input())
    if perfect(x):
        print(x,"là số hoàn hảo")
    else:
        print(x,"không phải số hoàn hảo")
except ValueError:
    print("Lỗi")
```

```python
#W5A4 (cách khác)
try:
    def perfect(a):
        return sum(i for i in range(1, a) if a % i == 0) == a
    x = int(input())
    print("Hoàn hảo" if perfect(x) else "Không hoàn hảo")
except ValueError:
    print("Lỗi")
```

```python
#W5A5
try:
    def find(lst,k):
        for i in range(len(lst)):
            if lst[i]==k:
                return i
        return -1
    ds=[1,2,3,4,5,6,7]
    k = int(input())
    print(find(ds,k))
except ValueError:
    print("Lỗi")
```

```python
#W5A6
def giai_thua(n):
    if n < 0:
        return False
    tich = 1
    for i in range(1, n + 1):
        tich *= i
    return tich
try:
    n = int(input())
    print(f"{n}! =", giai_thua(n))
except ValueError:
    print("Lỗi")
```

```python
#W5A7
def may_tinh(num1, num2, operat):
    if operat == '+':
        ket_qua = num1 + num2
    elif operat == '-':
        ket_qua = num1 - num2
    elif operat == '*':
        ket_qua = num1 * num2
    elif operat == '/':
        if num2 == 0:
            return "Không chia được cho 0"
        ket_qua = num1 / num2
    else:
        return "Toán tử không hợp lệ"
    return round(ket_qua, 2)
print(may_tinh(5.5, 2.3, '+'))
```

```python
#W5A8
def hamming(x, y):
    return bin(x ^ y).count('1')
print(hamming(1, 4)) 
```

```python
#W5A9
def tong_chu_so(n):
    return sum(int(ch) for ch in str(n))
print(tong_chu_so(1234)) 
```

```python
#W5A10
def dang_cau(a, b):
    if len(a) != len(b):
        return False

    map_a_to_b = {}
    map_b_to_a = {}

    for ch1, ch2 in zip(a, b):
        if ch1 in map_a_to_b:
            if map_a_to_b[ch1] != ch2:
                return False
        else:
            map_a_to_b[ch1] = ch2

        if ch2 in map_b_to_a:
            if map_b_to_a[ch2] != ch1:
                return False
        else:
            map_b_to_a[ch2] = ch1

    return True
```
