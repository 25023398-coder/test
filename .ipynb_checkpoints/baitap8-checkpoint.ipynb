```python
#W7A17
def majority_element(arr):
    candidate, count = None, 0
    for num in arr:
        if count == 0:
            candidate, count = num, 1
        elif num == candidate:
            count += 1
        else:
            count -= 1
    if arr.count(candidate) > len(arr)//2:
        return candidate
    return None
arr = list(map(int,input().split()))
print(majority_element(arr))
```

```python
#W7A18
def sort_even_odd(arr):
    evens = sorted([x for x in arr if x % 2 == 0])
    odds = sorted([x for x in arr if x % 2 == 1], reverse=True)
    return evens + odds
```

```python
#W7A20
def sum_k_digits(arr, k):
    nums = [x for x in arr if len(str(x)) == k]
    return sum(nums)
```

```python
#W7A22
def most_common_name(names):
    freq = Counter(names)
    max_count = max(freq.values())
    candidates = [name for name, c in freq.items() if c == max_count]
    return min(candidates)
```

```python
#W7A24
def remove_adjacent_duplicates(text):
    words = text.split()
    result = []
    for w in words:
        if not result or result[-1] != w:
            result.append(w)
    return " ".join(result)
```

```python
#W7A26
def sum_and_avg(text):
    nums = list(map(int, re.findall(r"\d+", text)))
    total = sum(nums)
    avg = total / len(nums) if nums else 0
    return total, avg
```

```python
#W7A27
def valid_parentheses(s):
    stack = []
    pairs = {')':'(', ']':'[', '}':'{'}
    for ch in s:
        if ch in "([{":
            stack.append(ch)
        elif ch in ")]}":
            if not stack or stack[-1] != pairs[ch]:
                return "NO"
            stack.pop()
    return "YES" if not stack else "NO"
```
