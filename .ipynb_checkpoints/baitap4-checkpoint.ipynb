```python
#W4A6
try:
    a = int(input())
    b = int(input())
    if a <= 0 or b <= 0 or a > b:
        print("a, b không hợp lệ")
    else:
        tong = 0
        for n in range(a, b + 1):
            if n < 2:
                continue
            flag = True
            for i in range(2, int(n**0.5)+1):
                if n % i == 0:
                    flag = False
                    break
            if flag:
                tong += n
        print("Tổng các số nguyên tố trong [{}, {}] là:{}".format(a, b, tong))
except ValueError:
    print("Lỗi")
```

```python
# cách khác
a = int(input())
b = int(input())
tong = 0
x = a
while x <= b:
    if x > 1:
        i = 2
        la_snt = True
        while i * i <= x:
            if x % i == 0:
                la_snt = False
                break
            i += 1
        if la_snt:
            tong += x
    x += 1
print(tong)
```

```python
#WA11
try:
    n = int(input("Số nguyên dương n:"))
    if n <= 0 or n >= 10**6:
        print("Sai đkien")
    else:
        dem = 0
        for i in range(1, n + 1):
            if n % i == 0 and i % 2 == 0:
                dem += 1
        print("Số lượng ước chẵn của", n, "là:", dem)
except ValueError:
    print("Lỗi")
```

```python
#WA12
try:
    A = int(input("Tiền gửi vào:"))
    n = int(input("Số tháng gửi:"))
    
    if A <= 0 or n <= 0:
        print("Không hợp lệ")
    else:
        for i in range(n):
            A += A * 0.007  
        print("Tiền nhận được:", int(A))  
except ValueError:
    print("Lỗi")
```

```python
#WA13
try:
    a = int(input())
    b = int(input())
    if a <= 0 or b <= 0:
        print("false")
    else:
        tong_a = 0
        for i in range(1, a):
            if a % i == 0:
                tong_a += i
        tong_b = 0
        for i in range(1, b):
            if b % i == 0:
                tong_b += i
        if tong_a == b and tong_b == a:
            print("true")
        else:
            print("false")
except ValueError:
    print("Lỗi")
```

```python
# cách khác
a = int(input())
b = int(input())
tong_a = 0
i = 1
while i < a:
    if a % i == 0:
        tong_a += i
    i += 1
tong_b = 0
j = 1
while j < b:
    if b % j == 0:
        tong_b += j
    j += 1
if tong_a == b and tong_b == a:
    print("true")
else:
    print("false")
```

```python
#WA14
try:
    m = int(input())
    n = int(input())
    if m <= 0 or n <= 0:
        print("Sai đkien")
    else:
        ketqua = 1
        for i in range(1, min(m, n) + 1):
            if m % i == 0 and n % i == 0:
                ketqua = i
        print("Ước chung lớn nhất:", ketqua)
except ValueError:
    print("Lỗi")
```

```python
#WA15
try:
    tongcon = int(input())
    tongchan = int(input())
    if tongcon <= 0 or tongchan <= 0:
        print("invalid")
    else:
        x = (4*tongcon - tongchan)//2
        y = tongcon - x
        if x >= 0 and y >= 0 and 2 * x + 4 * y == tongchan:
            print("Số gà:", x)
            print("Số chó:", y)
        else:
            print("invalid")
except ValueError:
    print("Lỗi")
```

```python
# cách khác
tong_con = int(input())
tong_chan = int(input())
ga = 0
find = False
while ga <= tong_con:
    cho = tong_con - ga
    chan = 2 * ga + 4 * cho
    if chan == tong_chan:
        print(ga)
        print(cho)
        find = True
        break
    ga += 1
if not find:
    print("Lỗi")
```
