```python
#W11A1
def reverse(path:str) -> int:
    with open(path, "r") as f:
        nums = f.read().strip()
    return int(nums[::-1])
print(reverse("input.txt"))
```

```python
#W11A2
def canWinNim(path:str) -> bool:
    with open(path, 'r') as f:
        n = int(f.read().strip())
    return n%4 != 0
print(canWinNim('input.txt'))
```

```python
#W11A3
def Hogwart(path:str):
    with open(path, 'r') as f:
        lines = f.readlines()
        n = int(lines[0].strip())
        for i in range(1,n+1):
            print(lines[i].strip())
Hogwart("Gryffindor.txt")
```

```python
#W11A5
def count(path:str, character:str):
    with open(path, 'r') as f:
        texts = f.read()
    return texts.count(character)
print(count('input.txt','e'))
```

```python
#W11A6
def countCharacter(path:str, letter:str) -> int:
    with open(path, 'r') as f:
        texts = f.read().lower()
    return texts.count(letter.lower())
print(countCharacter('input.txt','e'))

with open('input.txt') as f:
    text = f.read()
print(text)
print(text.lower().count('e'))
```
