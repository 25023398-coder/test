```python
#Viết một chương trình để đếm số từ trong một câu nhập từ người dùng.
def demsotu():
    cau = input()
    tach = cau.split()
    sotu = len(tach)
    print(sotu)
demsotu()
```

```python
#Viết một chương trình tính tổng n số đầu tiên trong dãy Fibonacci
def tongfibonacci(n):
    if n < 0:
        return "loi"
    a, b = 0, 1
    tong = 0
    for _ in range(n):
        a, b = b, b+a
        tong += a
    return tong
n = int(input())
print(tongfibonacci(n))
#câu này sai vì cập nhập số trước mới tính tổng
```
