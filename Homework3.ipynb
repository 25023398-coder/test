```python
#bài4
h = int(input())
i = 1
while i < h:
    print(" " * (h - i), end="")
    print("*", end="")
    if i > 1:
        print(" " * (2 * i - 3), end="")
        print("*")
    else:
        print()
    i += 1
print("*" * (2 * h - 1))
```

```python
#bài7
a = int(input("Nhập a: "))
dem = 0
for i in range(1, a+1):
    if a % i == 0:
        dem += 1
print("Số ước của a =", dem)
```

```python
#bài8
a = int(input("Nhập a: "))
b = int(input("Nhập b: "))

print("Ước chung của a và b:")
for i in range(1, min(a, b)+1):
    if a % i == 0 and b % i == 0:
        print(i, end=" ")
```
